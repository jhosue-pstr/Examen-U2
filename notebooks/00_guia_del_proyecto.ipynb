# Guía rápida · Proyecto ENAHO–ESCALE 2025

## ¿Quieres ejecutar todo de una sola vez?

Desde la raíz del proyecto, abre PowerShell y ejecuta:

```powershell
.\.venv\Scripts\python.exe pipeline.py
```

Ese comando comprueba los datos, integra ENAHO con ESCALE, valida la base, explora variables, prueba clusters, entrena cuatro clasificadores y genera evaluación y SHAP. No tienes que ejecutar los notebooks para que el pipeline funcione.

## Orden y notebooks por etapa

1. **Preparar e integrar** → `01_integracion_datos.ipynb`: filtrar 2025, preservar llaves y unir persona, hogar, Sumaria y ESCALE.
2. **Explorar** → `02_exploracion_datos.ipynb`: revisar cohortes, faltantes y distribución de la etiqueta.
3. **Descubrir perfiles** → `03_entrenamiento_clustering.ipynb`: entrenar K-Means y Ward y comparar métricas.
4. **Predecir** → `04_entrenamiento_clasificacion.ipynb`: probar baseline y modelos avanzados con separación por conglomerado.
5. **Interpretar** → `05_explicabilidad_shap.ipynb`: revisar evaluación, SHAP global, dependencia y casos locales.
6. **Ver todo junto** → `06_pipeline_completo.ipynb`: ejecuta el orquestador y muestra productos de todas las etapas.

Cada notebook de etapa sirve para estudiar o repetir esa parte por separado. El comando `pipeline.py` es el que ejecuta todo automáticamente.
