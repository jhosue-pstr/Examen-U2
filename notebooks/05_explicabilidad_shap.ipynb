# 05 · Evaluación final y explicabilidad SHAP

Este notebook revisa curvas ROC/PR, calibración y matriz de confusión del modelo seleccionado, además de sus explicaciones SHAP globales, de dependencia y locales. Para generar salidas nuevas, ejecuta primero el notebook 04 o `python pipeline.py`.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pipeline.py").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("No se encontró pipeline.py.")
OUT = ROOT / "informes" / "resultados_clasificacion"
comparison = OUT / "comparacion_modelos_prueba.csv"
if not comparison.is_file():
    raise FileNotFoundError("Faltan resultados de clasificación; ejecuta 04 o pipeline.py.")
results = pd.read_csv(comparison)
selected = results.loc[results["seleccionado_por_cv"], "modelo"].tolist()
if len(selected) != 1:
    raise ValueError("Debe haber exactamente un modelo ganador.")
MODEL_DIR = OUT / selected[0]
print("Modelo seleccionado:", selected[0])
display(results)
```

## Métricas visuales y SHAP

Beeswarm resume contribuciones globales; dependence explora cómo cambian las contribuciones; waterfall explica dos predicciones locales. Ninguno establece causalidad.

```python
figures = [
    "curva_roc.png", "curva_precision_recall.png", "curva_calibracion.png",
    "matriz_confusion.png", "shap_beeswarm.png",
    "shap_dependencia_lengua_internet.png", "shap_dependencia_internet_lengua.png",
    "shap_waterfall_alto_riesgo.png", "shap_waterfall_bajo_riesgo.png",
]
for filename in figures:
    path = MODEL_DIR / filename
    if not path.is_file():
        raise FileNotFoundError(f"Falta {path}; vuelve a ejecutar la clasificación.")
    print(filename)
    display(Image(filename=str(path)))
```
